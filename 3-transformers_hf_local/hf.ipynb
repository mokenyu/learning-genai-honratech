```python
import torch
from diffusers import DiffusionPipeline
from huggingface_hub import login

# Log in to Hugging Face Hub to avoid unauthenticated request warnings
# You will be prompted to enter your token.
login()

# switch to "mps" for apple devices
pipe = DiffusionPipeline.from_pretrained("stable-diffusion-v1-5/stable-diffusion-v1-5", dtype=torch.bfloat16, device_map="cuda")

prompt = "Astronaut in a jungle, cold color palette, muted colors, detailed, 8k"
image = pipe(prompt).images[0]
```

```python
import torch
from PIL import Image
from transformers import AutoModelForCausalLM
model = AutoModelForCausalLM.from_pretrained(
    "tiiuae/Falcon-OCR",
    trust_remote_code=True,
    torch_dtype=torch.bfloat16,
    device_map="auto",
)
image = Image.open("frontend.png")
texts = model.generate(image)  # default category is "plain"
print(texts[0])
```
