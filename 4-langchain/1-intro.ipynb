```python
import langchain
langchain.__version__ # wrapper
```

```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["GEMINI_API_KEY"] = os.getenv("GEMINI_API_KEY")
```

```python
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """Get the weather of a prompted city"""
    return f"The weather in {city} is rainy."

agent = create_agent(
    model="google_genai:gemini-3.6-flash", # gemini-2.5-flash
    tools=[get_weather],
    system_prompt="You are a weather-teller."
)
agent
```

```python
### running the agent
response = agent.invoke({"messages": [{"role": "user", "content": "what is the wheather like in new york?"}]})
response
```
