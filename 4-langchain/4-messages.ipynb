### Messages

Important object definitions for messages and fundamental unit if context for models in Langchain. They contain role, content and metadata. "role" defines the type(AIMessage, ToolMessage)

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GEMINI_API_KEY"]=os.getenv("GEMINI_API_KEY")
model = init_chat_model("google_genai:gemini-3.5-flash")
```

##### Text Prompts vs Message Prompts

Text prompts are generally just strings; they ae used for straigtforward generation when retaining convo history is not priority.

Message prompts can be used in lenghty conversations as they handle convo history but they include a bit of code complexity. A list of messages passed as message objects is required.

```python
# Text Prompt

model.invoke("Hello")
```

Message Types: System Message, Human Message, AI Message, Tool Message

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages=[
    SystemMessage("You are a world-renounded poetry writer, flawless writer."),
    HumanMessage(content="wRITE A POEM ABOUT PARROTIC CARROTS", name="Harry de Souza", id="agent47")
]

response = model.invoke(messages)
response.content
```

```python
from langchain.messages import ToolMessage

ai_msg = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "San Francisco"},
        "id": "call_123"
    }]
)

weather_result = "Sunny, 72F"
tool_msg = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"
)

messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_msg,
    tool_msg
]

response = model.invoke(messages)
messages.append(response)

print(response.content)
print(messages)
print(response.usage_metadata)
```
